```python
#C039
E = list(input())
count_ten = 0
count = 0
for i in range(len(E)):
    if E[i] == "<":
        count_ten += 10
    elif E[i] == "/":
        count += 1
    else :
        pass
sum = count_ten + count
print(sum)
```
